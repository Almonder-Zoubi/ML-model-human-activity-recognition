```python
import pandas as pd

# comparing (activity) feature of 4 csv files
df = pd.read_csv('../Predictions/stack_predictions.csv', delimiter=';')
df2 = pd.read_csv('../VSC_Preprocess_Models&Compare/adaboost.csv', delimiter=';')
df3 = pd.read_csv('../Predictions/Model_Predictions.csv', delimiter=';')
df4 = pd.read_csv('../Predictions/xgboost_predictions.csv', delimiter=';')
df5 = pd.read_csv('../Predictions/randomforest_predictions.csv', delimiter=';')
df6 = pd.read_csv('../Predictions/neuralnetwork_predictions.csv', delimiter=';')
df7 = pd.read_csv('../Predictions/my_prediction.csv', delimiter=';')

# comparing these csv files
# comparing just activity label of the files
# finding how many different are ther between each two
# and how many are the same
# and how many are the same between all of them
# and how many are different between all of them

# comparing stack_predictions.csv and DL_Model_Prediction_Process.csv
# store results in dictionary
same = {"stack_AdaB": 0, "stack_Model":0, "stack_xgb":0, "stack_rf":0, "stack_NN":0, "stack_CV":0, "AdaB_Model":0, "AdaB_xgb":0, "AdaB_rf":0, "AdaB_NN":0, "AdaB_CV":0, "Model_xgb":0, "Model_rf":0, "Model_NN":0, "Model_CV":0, "xgb_rf":0, "xgb_NN":0, "xgb_CV":0, "rf_NN":0, "rf_CV":0, "NN_CV":0, "all":0}
diff = {"stack_AdaB": 0, "stack_Model":0, "stack_xgb":0, "stack_rf":0, "stack_NN":0, "stack_CV":0, "AdaB_Model":0, "AdaB_xgb":0, "AdaB_rf":0, "AdaB_NN":0, "AdaB_CV":0, "Model_xgb":0, "Model_rf":0, "Model_NN":0, "Model_CV":0, "xgb_rf":0, "xgb_NN":0, "xgb_CV":0, "rf_NN":0, "rf_CV":0, "NN_CV":0, "all":0}
for i in range(len(df)):
    if df['activity'][i] == df2['prediction(activity)'][i]:
        same['stack_AdaB'] += 1
    else:
        diff['stack_AdaB'] += 1

# comparing stack_predictions.csv and Model_Predictions.csv
for i in range(len(df)):
    if df['activity'][i] == df3['prediction(activity)'][i]:
        same['stack_Model'] += 1
    else:
        diff['stack_Model'] += 1

# comparing stack_predictions.csv and xgboost_predictions.csv
for i in range(len(df)):
    if df['activity'][i] == df4['activity'][i]:
        same['stack_xgb'] += 1
    else:
        diff['stack_xgb'] += 1

# comparing stack_predictions.csv and randomforest_predictions.csv
for i in range(len(df)):
    if df['activity'][i] == df5['activity'][i]:
        same['stack_rf'] += 1
    else:
        diff['stack_rf'] += 1

# comparing stack_predictions.csv and neuralnetwork_predictions.csv
for i in range(len(df)):
    if df['activity'][i] == df6['activity'][i]:
        same['stack_NN'] += 1
    else:
        diff['stack_NN'] += 1

# comparing stack_predictions.csv and cross_validation_predictions.csv
for i in range(len(df)):
    if df['activity'][i] == df7['prediction(activity)'][i]:
        same['stack_CV'] += 1
    else:
        diff['stack_CV'] += 1

# comparing DL_Model_Prediction_Process.csv and Model_Predictions.csv
for i in range(len(df)):
    if df2['prediction(activity)'][i] == df3['prediction(activity)'][i]:
        same['AdaB_Model'] += 1
    else:
        diff['AdaB_Model'] += 1

# comparing DL_Model_Prediction_Process.csv and xgboost_predictions.csv
for i in range(len(df)):
    if df2['prediction(activity)'][i] == df4['activity'][i]:
        same['AdaB_xgb'] += 1
    else:
        diff['AdaB_xgb'] += 1

# comparing DL_Model_Prediction_Process.csv and randomforest_predictions.csv
for i in range(len(df)):
    if df2['prediction(activity)'][i] == df5['activity'][i]:
        same['AdaB_rf'] += 1
    else:
        diff['AdaB_rf'] += 1

# comparing DL_Model_Prediction_Process.csv and neuralnetwork_predictions.csv
for i in range(len(df)):
    if df2['prediction(activity)'][i] == df6['activity'][i]:
        same['AdaB_NN'] += 1
    else:
        diff['AdaB_NN'] += 1

# comparing DL_Model_Prediction_Process.csv and cross_validation_predictions.csv
for i in range(len(df)):
    if df2['prediction(activity)'][i] == df7['prediction(activity)'][i]:
        same['AdaB_CV'] += 1
    else:
        diff['AdaB_CV'] += 1

# comparing Model_Predictions.csv and xgboost_predictions.csv
for i in range(len(df)):
    if df3['prediction(activity)'][i] == df4['activity'][i]:
        same['Model_xgb'] += 1
    else:
        diff['Model_xgb'] += 1

# comparing Model_Predictions.csv and randomforest_predictions.csv
for i in range(len(df)):
    if df3['prediction(activity)'][i] == df5['activity'][i]:
        same['Model_rf'] += 1
    else:
        diff['Model_rf'] += 1

# comparing Model_Predictions.csv and neuralnetwork_predictions.csv
for i in range(len(df)):
    if df3['prediction(activity)'][i] == df6['activity'][i]:
        same['Model_NN'] += 1
    else:
        diff['Model_NN'] += 1

# comparing Model_Predictions.csv and cross_validation_predictions.csv
for i in range(len(df)):
    if df3['prediction(activity)'][i] == df7['prediction(activity)'][i]:
        same['Model_CV'] += 1
    else:
        diff['Model_CV'] += 1

# comparing xgboost_predictions.csv and randomforest_predictions.csv
for i in range(len(df)):
    if df4['activity'][i] == df5['activity'][i]:
        same['xgb_rf'] += 1
    else:
        diff['xgb_rf'] += 1

# comparing xgboost_predictions.csv and neuralnetwork_predictions.csv
for i in range(len(df)):
    if df4['activity'][i] == df6['activity'][i]:
        same['xgb_NN'] += 1
    else:
        diff['xgb_NN'] += 1

# comparing xgboost_predictions.csv and cross_validation_predictions.csv
for i in range(len(df)):
    if df4['activity'][i] == df7['prediction(activity)'][i]:
        same['xgb_CV'] += 1
    else:
        diff['xgb_CV'] += 1

# comparing randomforest_predictions.csv and neuralnetwork_predictions.csv
for i in range(len(df)):
    if df5['activity'][i] == df6['activity'][i]:
        same['rf_NN'] += 1
    else:
        diff['rf_NN'] += 1

# comparing randomforest_predictions.csv and cross_validation_predictions.csv
for i in range(len(df)):
    if df5['activity'][i] == df7['prediction(activity)'][i]:
        same['rf_CV'] += 1
    else:
        diff['rf_CV'] += 1

# comparing neuralnetwork_predictions.csv and cross_validation_predictions.csv
for i in range(len(df)):
    if df6['activity'][i] == df7['prediction(activity)'][i]:
        same['NN_CV'] += 1
    else:
        diff['NN_CV'] += 1

# comparing all of them
# for i in range(len(df)):
#     if df['activity'][i] == df2['prediction(activity)'][i] == df3['prediction(activity)'][i] == df4['activity'][i] == df5['activity'][i] == df6['activity']:
#         same['all'] += 1
#     else:
#         diff['all'] += 1
# print('All of them')
# print('Same:', same['all'])
# print('Different:', diff['all'])

# Perform element-wise comparisons and combine them using logical AND
all_same = (df['activity'] == df2['prediction(activity)']) & \
           (df['activity'] == df3['prediction(activity)']) & \
           (df['activity'] == df4['activity']) & \
           (df['activity'] == df5['activity']) & \
           (df['activity'] 
           == df6['activity'])

# Count the number of True and False values
same['all'] = all_same.sum()
diff['all'] = len(df) - same['all']

# print same from higher to lower and diff lower to higer
sorted_same = {k: v for k, v in sorted(same.items(), key=lambda item: item[1])}
print(sorted_same)
sorted_diff = {k: v for k, v in sorted(diff.items(), key=lambda item: item[1], reverse=True)}
print(sorted_diff)

```
