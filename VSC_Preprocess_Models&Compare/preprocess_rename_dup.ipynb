```python
import pandas as pd

# Load the data
df = pd.read_csv('../to_predict.csv', delimiter=';')

# Convert string numbers with commas to floats
for column in df.columns:
    if df[column].dtype == 'object':
        try:
            df[column] = df[column].str.replace(',', '.').astype(float)
        except ValueError:
            pass  # If conversion fails, it's probably a categorical column

# saving the first row (features row) in a seprated list
features = df.columns

# find the duplicated columns put them in dictionary with their indexes
# the moment you finde duplicate rename it make it unique by just appending suffix
# the suffix can be the index of the column
# save the new name in a dictionary
# rename the columns in the dataframe
# save the csv file with the new unique names
duplicated_columns = {}
for i, column in enumerate(df.columns):
    if column in df.columns[:i]:
        duplicated_columns[column] = duplicated_columns.get(column, []) + [i]
        df.rename(columns={column: f'{column}_{i}'}, inplace=True)

# rename the columns in the dataframe
df.columns = features
# after renaming columns save the csv file with the new unique names
df.to_csv('../new/edited2_to_predict.csv', sep=';', index=False)
```
